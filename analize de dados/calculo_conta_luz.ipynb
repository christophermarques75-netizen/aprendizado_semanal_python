```python
import pandas as pd


df = pd.read_csv('base_dados_contaluz.csv')


colunas_essenciais = [
    'consumo_kwh',      # Consumo ativo medido no período
    'tarifa_energia',    # Valor da tarifa base por kWh (TE + TUSD)
    'bandeira_tarifaria',# Se houver custo adicional por bandeira (Amarela, Vermelha)
    'demanda_contratada',# Crucial para contas EMPRESARIAIS (Grupo A)
    'pis_percentual',    # Imposto federal
    'cofins_percentual', # Imposto federal
    'icms_percentual',   # Imposto estadual
    'cosip_ilum_publica' # Contribuição de iluminação pública (comum em residências)
]

# 3. Filtrar o DataFrame mantendo APENAS as colunas que agregam valor ao cálculo
# O parâmetro 'errors="ignore"' garante que o código não quebre se alguma coluna não existir
df_limpo = df[df.columns.intersection(colunas_essenciais)].copy()

# 4. Exemplo de remoção explícita de colunas inúteis (caso prefira a lógica inversa)
# Colunas como IDs, nomes, datas de emissão ou endereços não alteram o valor do cálculo
colunas_para_eliminar = [
    'id_cliente', 'nome_cliente', 'endereco', 'numero_medidor', 
    'data_emissao', 'data_vencimento', 'status_pagamento'
]
df_limpo = df_limpo.drop(columns=[c for c in colunas_para_eliminar if c in df_limpo.columns])

# 5. Salvar o novo CSV limpo
df_limpo.to_csv('conta_luz_limpo.csv', index=False)

print("CSV limpo com sucesso! Colunas restantes:")
print(df_limpo.columns.tolist())
```
